```python
#functions of largest of two numbers
def large_number(a,b):
    if(a > b):
        return a
    else:
        return b
result = large_number(10,20)
print("Largest:",result)

    
```

```python
# function +loop
def multiplication_table(number):
    for i in range(1,11):
        print(number,"X",i,"=",number * i)
number = int(input("Enter number:"))
multiplication_table(number)
```

```python
#largest of three numbers
def largest(a,b,c):
    if(a>=b and a>=c):
        return a
    elif(b>=c and b>=c):
        return b
    else:
        return c
a = int(input("Enter a:"))
b=int(input("Enter b"))
c=int(input("Enter c:"))
print("Largest=",largest(a,b,c))
```

```python
# grade calculation
def find_grade(marks):
    if(marks>=90):
        return "A"
    elif(marks >=80):
        return "B"
    elif(marks >=70):
        return "C"
    elif(marks>=60):
        return"D"
    elif(marks>=40):
        return "E"
    else:
        return "Fail"
marks=int(input("Enter marks:"))
print("Grade=",find_grade(marks))
```
